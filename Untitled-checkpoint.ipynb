```python
**Detection of outliers**
```

```python
for column in ["Âge", "Salaire"]:
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]

    print(f"Outliers in {column}:")
    print(outliers)
```

```python
for column in ["Âge", "Salaire"]:
    plt.figure(figsize=(6, 4))
    sns.boxplot(y=data[column])
    plt.title(f"Box Plot of {column}")
    plt.ylabel(column)
    plt.show()
```

**Do salaries differ across occupations?**

```python
data.groupby("Profession")["Salaire"].mean().sort_values(ascending=False)
```

```python
avg_salary = (
    data.groupby("Profession")["Salaire"]
        .mean()
        .sort_values(ascending=False)
)


plt.figure(figsize=(8, 5))
sns.barplot(
    x=avg_salary.index,
    y=avg_salary.values,
    hue=avg_salary.values,
    palette="Blues"
)

plt.title("Average Salary by Profession")
plt.xlabel("Profession")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Do salaries differ across cities?**

```python
data.groupby("Ville")["Salaire"].mean().sort_values(ascending=False)
```

```python
avg_salary_vil = (
    data.groupby("Ville")["Salaire"].mean().sort_values(ascending=False))


plt.figure(figsize=(8, 5))
sns.barplot(
    x=avg_salary_vil.index,
    y=avg_salary_vil.values,
    hue=avg_salary_vil.values,
    palette="viridis"
)

plt.title("Average Salary by city")
plt.xlabel("Profession")
plt.ylabel("Average Salary")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
